```python
# aplicação de mascara
cpf = input ('informe o cpf')

#cpf tem 11 digitos
if len(cpf) == 11 and cpf.isdigit():
  cpf_mascarado = f"`{cpf[:3]}.{cpf[3:6]}.{cpf[6:9]}-{cpf[9:11]}"

  print(cpf_mascarado)

else:
  print('cpf informado esta errado')
```

```python
# ocultamento
cartao = input('informe o numero do cartão')

if len(cartao)==16 and cartao.isdigit():
  cartao_mascarado = f" **** **** **** {cartao[-4:]}"
  print(cartao_mascarado)
else:
  print('Numero de cartão invalido')
```
