```python
# conectar de api
import requests
import pandas as pd
# para coletar dados da api preciso do endpoint
url = 'https://api.open-meteo.com/v1/forecast?latitude=-23.090000&longitude=-47.218000&current_weather=true'

#vamos descobrir se a api esta funcionando
resposta = requests.get(url)
#status code 200 é ok
if resposta.status_code == 200:
  # é só chamar o que a gente precisa dos dados
  # geralmente dados de api vem no formato json
  dados_json = resposta.json()
  clima_atual = dados_json.get("current_weather")

  # usar pandas para exibir
  df_clima = pd.DataFrame([clima_atual])
  display(df_clima)
```

```python
# para usar uma versao não instalada do sql usamos
import sqlite3

#para ler um arquivo .sql
#deve-se criar uma conexão temporaria
conn = sqlite3.connect('memory:')
cursor = conn.cursor()

# o colab, temporariamente é nosso "workbench"
cursor.executescript('''
    create table teste(
    teste_id int primary key,
    nome_teste varchar(255) not null,
    numero_teste integer not null
    );
    ''')

conn.commit()
query_e = '''
select * from teste;
'''

#paraa ver os dados da query, devemos usar o pandas
df_query = pd.read_sql(query_e, conn)

display(df_query)
```
