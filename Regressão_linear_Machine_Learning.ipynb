```python
# blibliotecas para previsão de dados
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

#importar a base de dados
df_notas = pd.read_csv('/content/Copy of scatter_plot_trend_line_data.csv')

# faz a separação de var_independente e var_dependente
# var_independente = X
# var_dependente = Y

X = df_notas[['Hours Studied']]
Y = df_notas[['Test Score']]

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state=42)
# separamos nossas variaveis independentes e dependentes em treino e teste
# criar as var : X_train, X_teste, Y_train, Y_test
# o comando train_test_split(X, Y test_size = ----. random state=42)
# pega as independentes e as dependentes e escolhe uma porrcentagem de teste (-----= 0.x%)
model = LinearRegression()
model.fit(X_train, Y_train) # fit é o comando que faz a IA aprender

# a IA prrecisa saber o que ela vai prever
# vamos fazer ela testar os valores independentes
y_prev = model.predict(X_test)

horas = float(input('Quantas horas você estudou? '))
previsao = model.predict([[horas]])
print(f'a sua previsão de nota é {previsao[0]}')
```

## Visualizando a Regressão Linear

```python
plt.figure(figsize=(10, 6))
plt.scatter(X, Y, color='blue', label='Dados Reais')
plt.plot(X_test, y_prev, color='red', linewidth=2, label='Linha de Regressão Prevista')
plt.xlabel('Horas Estudadas')
plt.ylabel('Nota no Teste')
plt.title('Regressão Linear: Horas Estudadas vs. Nota no Teste')
plt.legend()
plt.grid(True)
plt.show()
```

### Comparação entre Valores Reais e Preditos

```python
plt.figure(figsize=(10, 6))
plt.scatter(Y_test, y_prev, color='green', label='Valores Previstos')
plt.plot([Y.min(), Y.max()], [Y.min(), Y.max()], 'k--', lw=2, label='Linha Ideal (y=x)')
plt.xlabel('Nota Real')
plt.ylabel('Nota Prevista')
plt.title('Notas Reais vs. Notas Previstas')
plt.legend()
plt.grid(True)
plt.show()
```
