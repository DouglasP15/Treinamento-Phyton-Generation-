```python
import pandas as pd
import plotly.express as px

df_prod_doces = pd.read_csv('/content/estatisticapotly/probabilidade - Base de Inspeção de Doces Individuais.csv')

fig_qtdSabor = px.histogram(
    df_prod_doces, x='Sabor',title = 'Quantidade de doces produzidos',
    color = "Sabor",
    text_auto=True
)
fig_qtdSabor.update_layout(
    title_x=0.5,
    yaxis_title="Quantidade"
)
fig_qtdSabor.show()
```

```python
# taxa de defeitos tem que mapear
df_prod_doces['Taxa de defeito'] = df_prod_doces['Defeito'].map({0:'Sem defeito',1:'Com defeito'})

fig_defeitos = px.pie(
    df_prod_doces,
    names="Taxa de defeito",
    title="Taxa de defeitos nos doces",
    color='Taxa de defeito',
    hole=0.5
)
fig_defeitos.update_layout(title_x=0.5)
fig_defeitos.show()
```
